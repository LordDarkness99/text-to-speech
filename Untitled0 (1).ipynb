```python
import os
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import wave
import contextlib
from tqdm import tqdm

dataset_path = '/content/drive/MyDrive/ColabNotebooks/ml/tts/LJSpeech-1.1'
csv_path = os.path.join(dataset_path, 'metadata.csv')
wav_dir = os.path.join(dataset_path, 'wavs')

print("Memuat dataset metadata...")
df = pd.read_csv(csv_path, sep='|', header=None, names=['id', 'text', 'normalized_text'])

print("\n--- Pengecekan Missing Values ---")
print(df.isnull().sum())
df['normalized_text'].fillna(df['text'], inplace=True)
df.dropna(inplace=True) # Buang jika ID atau Teks asli kosong

df['char_length'] = df['normalized_text'].apply(len)
df['word_count'] = df['normalized_text'].apply(lambda x: len(str(x).split()))

print("\nMengekstrak durasi audio (proses ini memakan waktu 5 JAMMMM)...")
durations = []
valid_ids = []

for file_id in tqdm(df['id']):
    wav_path = os.path.join(wav_dir, f"{file_id}.wav")
    try:
        with contextlib.closing(wave.open(wav_path, 'r')) as f:
            frames = f.getnframes()
            rate = f.getframerate()
            duration = frames / float(rate)
            durations.append(duration)
            valid_ids.append(file_id)
    except Exception as e:
        durations.append(None) # Jika file wav korup atau tidak ditemukan

df['audio_duration'] = durations

df = df.dropna(subset=['audio_duration'])

sns.set_theme(style="whitegrid")
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

sns.histplot(df['audio_duration'], bins=50, kde=True, ax=axes[0], color='skyblue')
axes[0].set_title('Distribusi Durasi Audio (Detik)')
axes[0].set_xlabel('Detik')

sns.histplot(df['word_count'], bins=50, kde=True, ax=axes[1], color='lightgreen')
axes[1].set_title('Distribusi Jumlah Kata per Kalimat')
axes[1].set_xlabel('Jumlah Kata')

sns.scatterplot(x='audio_duration', y='char_length', data=df, ax=axes[2], alpha=0.3, color='salmon')
axes[2].set_title('Korelasi Durasi Suara vs Panjang Teks')
axes[2].set_xlabel('Durasi (Detik)')
axes[2].set_ylabel('Panjang Karakter')

plt.tight_layout()
plt.show()

print("\n--- Ringkasan Statistik Dataset ---")
print(df[['audio_duration', 'word_count', 'char_length']].describe())
```

```python
import os
import torch
import pandas as pd
from datasets import Dataset, Audio
from dataclasses import dataclass
from typing import Any, Dict, List, Union
from transformers import (
    SpeechT5Processor,
    SpeechT5ForTextToSpeech,
    Seq2SeqTrainingArguments,
    Seq2SeqTrainer
)
```

```python
print("Mempersiapkan Dataset LJSpeech (FULL DATA)...")
dataset_path = '/content/drive/MyDrive/ColabNotebooks/ml/tts/LJSpeech-1.1'
csv_path = os.path.join(dataset_path, 'metadata.csv')

df = pd.read_csv(csv_path, sep='|', header=None, names=['id', 'text', 'normalized_text'])
df['normalized_text'] = df['normalized_text'].fillna(df['text'])
df = df.dropna()

# df = df.head(300)

df['audio'] = df['id'].apply(lambda x: os.path.join(dataset_path, 'wavs', f"{x}.wav"))

hf_dataset = Dataset.from_pandas(df[['audio', 'normalized_text']])
hf_dataset = hf_dataset.cast_column("audio", Audio(sampling_rate=16000))
```

```python
print("Memuat Processor dan Model SpeechT5...")
processor = SpeechT5Processor.from_pretrained("microsoft/speecht5_tts")
model = SpeechT5ForTextToSpeech.from_pretrained("microsoft/speecht5_tts")
model.config.use_cache = False # Wajib False saat training untuk menghemat memori
```

```python
print("Mengekstrak Spectrogram dan Teks (LAMA)")
speaker_embedding = torch.randn(1, 512)

def prepare_dataset(batch):
    audio = batch["audio"]
    inputs = processor(
        text=batch["normalized_text"],
        audio_target=audio["array"],
        sampling_rate=audio["sampling_rate"],
        return_attention_mask=False,
    )
    batch["input_ids"] = inputs.get("input_ids")
    batch["labels"] = inputs.get("labels")
    batch["speaker_embeddings"] = speaker_embedding[0]
    return batch

hf_dataset = hf_dataset.map(prepare_dataset, remove_columns=hf_dataset.column_names)
```

```python
@dataclass
class TTSDataCollatorWithPadding:
    processor: Any

    def __call__(self, features: List[Dict[str, Union[List[int], torch.Tensor]]]) -> Dict[str, torch.Tensor]:
        input_ids = [{"input_ids": feature["input_ids"]} for feature in features]
        label_features = [{"input_values": feature["labels"]} for feature in features]
        speaker_features = [feature["speaker_embeddings"] for feature in features]

        # Pad text
        batch = self.processor.pad(input_ids=input_ids, return_tensors="pt")
        # Pad spectrograms
        with self.processor.as_target_processor():
            labels_batch = self.processor.pad(labels=label_features, return_tensors="pt")

        labels = labels_batch["labels"].masked_fill(labels_batch.attention_mask.unsqueeze(-1) == 0, -100)
        batch["labels"] = labels
        batch["speaker_embeddings"] = torch.tensor(speaker_features)

        return batch

data_collator = TTSDataCollatorWithPadding(processor=processor)
```

```python
import torch
import torch.nn.functional as F
from transformers import Seq2SeqTrainer

class TTSDataCollatorWithPadding:
    def __init__(self, processor, model):
        self.processor = processor
        self.reduction_factor = model.config.reduction_factor

    def __call__(self, features):
        input_ids = []
        for feature in features:
            ids = feature["input_ids"]
            if isinstance(ids[0], list):
                ids = ids[0]
            input_ids.append({"input_ids": ids})

        batch = self.processor.pad(input_ids=input_ids, return_tensors="pt")

        labels = []
        for feature in features:
            lbl = torch.tensor(feature["labels"])
            if lbl.dim() == 3 and lbl.shape[0] == 1:
                lbl = lbl.squeeze(0)
            elif lbl.dim() == 2 and lbl.shape[0] == 80 and lbl.shape[1] != 80:
                lbl = lbl.transpose(0, 1)
            labels.append(lbl)

        labels_padded = torch.nn.utils.rnn.pad_sequence(
            labels, batch_first=True, padding_value=-100.0
        )

        max_len = labels_padded.shape[1]
        if max_len % self.reduction_factor != 0:
            pad_amount = self.reduction_factor - (max_len % self.reduction_factor)
            labels_padded = F.pad(labels_padded, (0, 0, 0, pad_amount), value=-100.0)

        batch["labels"] = labels_padded

        speaker_features = [torch.tensor(feature["speaker_embeddings"]) for feature in features]
        batch["speaker_embeddings"] = torch.stack(speaker_features)

        return batch

data_collator = TTSDataCollatorWithPadding(processor=processor, model=model)

print("Menyiapkan konfigurasi Full Training...")
training_args = Seq2SeqTrainingArguments(
    output_dir="/content/drive/MyDrive/ColabNotebooks/ml/tts/speecht5_finetuned",
    per_device_train_batch_size=4,
    gradient_accumulation_steps=8,
    learning_rate=1e-5,
    warmup_steps=100,
    max_steps=4000,
    gradient_checkpointing=True,
    fp16=True,
    eval_strategy="no",
    save_steps=500,
    logging_steps=50,
    report_to="none"
)

trainer = Seq2SeqTrainer(
    args=training_args,
    model=model,
    train_dataset=hf_dataset,
    data_collator=data_collator,
)

print("Memulai Fine-Tuning (Final Fix)!")
trainer.train()
```

```python
from transformers import SpeechT5ForTextToSpeech, SpeechT5Processor, SpeechT5HifiGan
import torch
from IPython.display import Audio, display

checkpoint_path = "/content/drive/MyDrive/ColabNotebooks/ml/tts/speecht5_finetuned/checkpoint-4000"

print("Memuat model hasil Full Fine-Tuning (4000 steps)...")
processor = SpeechT5Processor.from_pretrained("microsoft/speecht5_tts")
vocoder = SpeechT5HifiGan.from_pretrained("microsoft/speecht5_hifigan")
model_finetuned = SpeechT5ForTextToSpeech.from_pretrained(checkpoint_path)

teks_uji = "The architecture of machine learning models requires careful tuning and evaluation. After thousands of steps, the model should be able to speak fluently without stopping prematurely."
inputs = processor(text=teks_uji, return_tensors="pt")

torch.manual_seed(42)
speaker_embeddings = torch.randn(1, 512)

print("Menghasilkan gelombang suara...")
with torch.no_grad():
    speech = model_finetuned.generate_speech(inputs["input_ids"], speaker_embeddings, vocoder=vocoder)

print("\nSilakan dengarkan mahakarya Anda:")
display(Audio(speech.numpy(), rate=16000))
```

```python
# Install library untuk menghitung Word Error Rate
!pip install -q jiwer

from transformers import pipeline
import jiwer
import re

print("Memuat Juri Penilai (OpenAI Whisper)...")
asr_evaluator = pipeline("automatic-speech-recognition", model="openai/whisper-tiny")

audio_array = speech.numpy()

print("Mendengarkan dan mentranskripsi audio...")
hasil_transkripsi = asr_evaluator({"sampling_rate": 16000, "raw": audio_array})["text"]

def bersihkan_teks(teks):
    teks = teks.lower()
    teks = re.sub(r'[^a-z0-9\s]', '', teks)
    return teks.strip()

teks_asli = bersihkan_teks(teks_uji)
teks_terdengar = bersihkan_teks(hasil_transkripsi)

error_rate = jiwer.wer(teks_asli, teks_terdengar)
akurasi_persen = max(0, (1 - error_rate) * 100) # Memastikan akurasi tidak minus

print("\n" + "="*40)
print("HASIL EVALUASI OBJEKTIF MODEL TTS 🏆")
print("="*40)
print(f"Teks Target Asli  : '{teks_asli}'")
print(f"Didengar Whisper  : '{teks_terdengar}'")
print("-" * 40)
print(f"Tingkat Kesalahan (WER) : {error_rate * 100:.2f}%")
print(f"AKURASI PELAFALAN       : {akurasi_persen:.2f}%")
print("="*40)
```
